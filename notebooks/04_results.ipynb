# Reproduce the Headline Results
Reads pre-computed CSVs from `benchmarks/results/` and re-renders the main + ablation tables.

```python
import pandas as pd, matplotlib.pyplot as plt
main = pd.read_csv('../benchmarks/results/main_table.csv'); main
```

## Ablation bar chart

```python
abl = pd.read_csv('../benchmarks/results/ablation_table.csv')
abl.plot.bar(x='variant', y='macro_f1', legend=False)
plt.ylabel('Macro-F1'); plt.title('Ablation: Macro-F1 by variant')
```

## Certified-radius vs epsilon

```python
cert = pd.read_csv('../benchmarks/results/certification_radii.csv')
plt.plot(cert['epsilon'], cert['certified_fraction'], marker='o')
plt.xlabel('epsilon'); plt.ylabel('certified fraction')
```
