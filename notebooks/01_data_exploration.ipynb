# IIS3D Data Exploration
Load the IIS3D dataset, inspect class distribution and message-rate histograms.

```python
# TODO: configure data root
from mambaguard.data import DATASET_REGISTRY
ds = DATASET_REGISTRY['iis3d'](split='train', config={'root': 'data/raw/iis3d'})
len(ds)
```

## Class distribution

```python
from collections import Counter
import matplotlib.pyplot as plt
labels = [int(m.label) for m in ds]
c = Counter(labels)
plt.bar(list(c.keys()), list(c.values()))
plt.xlabel('class'); plt.ylabel('count'); plt.title('IIS3D class distribution')
```

## Message-rate histogram

```python
# TODO: bin timestamps into windows and plot rate
import numpy as np
ts = np.array([m.timestamp for m in ds])
plt.hist(np.diff(np.sort(ts)), bins=50)
plt.xlabel('inter-message dt (s)'); plt.ylabel('count')
```
