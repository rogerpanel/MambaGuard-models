# Certification Walk-Through
Compute the per-layer Lipschitz bound L_f, solve the Stackelberg game, then plot certified radius vs epsilon.

```python
import torch, numpy as np, matplotlib.pyplot as plt
from mambaguard.models import MambaGuard
from mambaguard.certification import compute_lipschitz_bound, StackelbergSolver, composed_certificate
ckpt = torch.load('outputs/run/best.ckpt', map_location='cpu')
model = MambaGuard.from_config(ckpt['config']); model.load_state_dict(ckpt['model'])
L_f = float(compute_lipschitz_bound(model)); L_f
```

## Stackelberg solve

```python
U = np.array([[0.95,0.80,0.75,0.85,0.10],[0.10,0.20,0.15,0.10,1.0],[0.90,0.85,0.80,0.85,0.40],[0.70,0.65,0.60,0.70,0.80],[0.85,0.75,0.80,0.85,0.30]])
sol = StackelbergSolver(utility=U).solve(); sol
```

## Certified radius vs epsilon

```python
eps = np.linspace(0.005, 0.15, 30)
bounds = [float(composed_certificate(v_star=sol['value'], L_f=L_f, epsilon=e, B=1.0, T=10000, n_defender_actions=5)) for e in eps]
plt.plot(eps, bounds); plt.xlabel('epsilon'); plt.ylabel('certified lower bound')
```
