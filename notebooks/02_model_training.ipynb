# MambaGuard Training Demo
Build a tiny MambaGuard and run a few training steps on a small subset.

```python
import yaml
from mambaguard.models import MambaGuard
cfg = yaml.safe_load(open('../configs/default.yaml'))
cfg['model']['d_model'] = 64
cfg['model']['n_layers'] = 2
model = MambaGuard.from_config(cfg)
sum(p.numel() for p in model.parameters())
```

## Build optimiser + trainer

```python
# TODO: load a small dataset slice
from mambaguard.training import MambaGuardTrainer, build_optimizer, build_scheduler
optim = build_optimizer(model, cfg)
sched = build_scheduler(optim, cfg)
trainer = MambaGuardTrainer(model=model, optimizer=optim, scheduler=sched, config=cfg, output_dir='outputs/demo')
```
